```python
import ast, glob, json
from collections import Counter

bench = json.load(open(glob.glob("/kaggle/input/**/benchmark_release.json", recursive=True)[0], encoding="utf-8"))

def values(e):
    o = e["options_list"]
    o = ast.literal_eval(o) if isinstance(o, str) else o
    return [float(str(v).rstrip("%")) for v in o]

for task in ["Building Damage Counting", "Road Damage Counting"]:
    qs = [e for e in bench if e.get("task") == task and e.get("post_image_type") == "Optical"]
    middle = rule = big = big_middle = 0
    by_rank = Counter()
    for e in qs:
        v = values(e)
        s = sorted(v)
        truth = v["ABCDE".index(e["ground_truth_option"].strip()[0])]
        by_rank[s.index(truth)] += 1
        middle += truth == s[2]
        rule += truth == (s[2] if s[4] > 20 else s[1])
        if s[4] > 20:
            big += 1
            big_middle += truth == s[2]
    n = len(qs)
    print(task, "| questions:", n)
    print("  always the middle value: %.2f%%" % (100 * middle / n))
    print("  training-set rule: %.2f%% (fixed on building counting only)" % (100 * rule / n))
    print("  largest option above 20: %d questions, middle value right on %d" % (big, big_middle))
    print("  rank of the correct value, smallest = 0:", sorted(by_rank.items()))
```
